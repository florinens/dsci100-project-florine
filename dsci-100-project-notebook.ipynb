Visualization 1: This chart compares average conference win percentage with final tournament finishes to test whether the team's conference record can predict how far a team advances. If it were a good predictor, averages would rise as the team's tourney finishes get better. Instead, the pattern is uneven with the top 16 losers average at 74% while top 8 losers average about 95%. This suggests conference win percentage alone is a weak predictor of tournament outcome would need to be used alongside seed as another predictor.

```python
finish_order = [
    "opening_round_loss", "first_round_loss", "second_round_loss",
    "top_16_loss", "top_8_loss", "top_4_loss", "top_2_loss", "champ"
]

conf_wins_plot = alt.Chart(wmm_tidy).mark_bar().encode(
    x = alt.X("tourney_finish:N").sort(finish_order).title("Tournament Finish"),
    y = alt.Y("mean(conf_wins_pct)").scale(zero=False).title("Average Conference Win Percentage")
).properties(
    title="Average Conference Win Percentage by Final Tournament Finish"
)

conf_wins_plot
```

Visualization 2: This chart compares average tournament seed with final tournament finishes to test whether the team's seed can predict how far a team advances. Average seed falls from about 10 for first round losers to about 2 to 3 for teams that reached the top 8 or beyond. Since a lower seed number would mean a better team, better seeds would tend to go further which would match the predicted pattern. This suggests that a team's seed is a strong predictor of tournament outcome, and a much clearer one than conference win percentage.

```python
seed_plot = alt.Chart(wmm_tidy).mark_bar().encode(
    x = alt.X("tourney_finish").sort(finish_order).title("Tournament Finish"),
    y = alt.Y("mean(seed)").title("Average Tournament Seed (1 = best)")
).properties(
    title="Average Tournament Seed by Final Tournament Finish"
)

seed_plot
```

Visualization 3: This chart compares average regional win percentage with final tournament finishes to test whether their regional win record can predict how far a team advances. Averages rise from about 76% for first round losers to about 90% for top 8 losers, which suggests stronger regular season teams tend to go further. However, the top 4, which is at about 84%, and top 2, which is at about 77%, averages drop back down, so the relationship is only partial. Regional win percentage is a moderate predictor, but it's weaker than seed.

```python
reg_wins_plot = alt.Chart(wmm_tidy).mark_bar().encode(
    x = alt.X("tourney_finish:N").sort(finish_order).title("Tournament Finish"),
    y = alt.Y("mean(reg_wins_pct)").scale(zero=False).title("Average Conference Win Percentage")
).properties(
    title="Average Regional Win Percentage by Final Tournament Finish"
)

reg_wins_plot
```

Tournament finish is categorical, so classification is needed. K-NN makes no assumptions about the shape of the relationship between predictors and outcome, so the uneven patterns in our plots are not a problem.

Predictors need to be standardised to ensure that they all have the same scale. Some tournament finish categories have less teams in those such as “champ” or “opening_round_loss” so KNN would rarely predict the outcome as those.

First, every seed of 0 will be re-coded as 8. The predictors will be standardized with a preprocessor fit on the training data only, inside a pipeline. I will use 5-fold cross-validation on the training set to choose a number of neighbours that is most accurate. The test set will not be used during training.
